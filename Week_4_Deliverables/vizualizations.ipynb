```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
print(df.describe())
```

```python
print(df.head(20))
print(df.info())
```

```python
# Line plot
df.sort_values('total_bill', inplace=True)
```

```python
plt.plot(df['total_bill'], df['tip'])
plt.title('Line Plot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
#scatterplot of size vs total bill
plt.scatter(df['size'], df['total_bill'])
plt.title('Scatter Plot of Size vs Total Bill')
plt.xlabel('Size')
plt.ylabel('Total Bill')
plt.show()
```

```python
#Histogram using seaborn
sns.histplot(df['total_bill'], bins=10, kde=True)
plt.title('Histogram of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.boxplot(x= df['smoker'], y= df['total_bill'])
plt.title('Box Plot of Total Bill by Smoker Status')
plt.xlabel('Smoker')
plt.ylabel('Total Bill')
plt.show()
```
